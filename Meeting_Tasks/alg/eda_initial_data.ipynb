```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_orp_1 = pd.read_csv(r'/Users/patrickwilson/Desktop/ChemE Repo/Chem-E-2025-2026/data/orp_data/recording1.csv')
df_orp_2 = pd.read_csv(r'/Users/patrickwilson/Desktop/ChemE Repo/Chem-E-2025-2026/data/orp_data/recording2.csv')
df_orp_3 = pd.read_csv(r'/Users/patrickwilson/Desktop/ChemE Repo/Chem-E-2025-2026/data/orp_data/recording3.csv')
```

```python
#standard deviation with rolling window of 10 data points

df_orp_1['std_10'] = df_orp_1['orp_mV'].rolling(window=10).std()
df_orp_2['std_10'] = df_orp_2['orp_mV'].rolling(window=10).std()
df_orp_3['std_10'] = df_orp_3['orp_mV'].rolling(window=10).std()

#average of 10 data points with rolling window
df_orp_1['avg_10'] = df_orp_1['orp_mV'].rolling(window=10).mean()
df_orp_2['avg_10'] = df_orp_2['orp_mV'].rolling(window=10).mean()
df_orp_3['avg_10'] = df_orp_3['orp_mV'].rolling(window=10).mean()

#standard deviation of the average with rolling window of 10 data points
df_orp_1['sta_10'] = df_orp_1['avg_10'].rolling(window=10).std()
df_orp_2['sta_10'] = df_orp_2['avg_10'].rolling(window=10).std()
df_orp_3['sta_10'] = df_orp_3['avg_10'].rolling(window=10).std()
```

```python
df_orp_1.describe()
df_orp_3["seconds"] = df_orp_3['timestamp_ms'] / 1000
df_orp_2["seconds"] = df_orp_2['timestamp_ms'] / 1000
df_orp_1["seconds"] = df_orp_1['timestamp_ms'] / 1000
```

```python
df_orp_3.plot(kind='scatter', 
              x='seconds', 
              y='orp_mV',
              title='mv over time reaction 3')


df_orp_2.plot(kind='scatter', 
              x='seconds', 
              y='orp_mV',
              title='mv over time reaction 2')
df_orp_1.plot(kind='scatter', 
              x='seconds', 
              y='orp_mV',
              title='mv over time reaction 1')


df_orp_3.plot(kind='scatter', 
              x='seconds', 
              y='std_10',
              title='std over time reaction 3')


df_orp_2.plot(kind='scatter', 
              x='seconds', 
              y='std_10',
              title='std over time reaction 2')
df_orp_1.plot(kind='scatter', 
              x='seconds', 
              y='std_10',
              title='std over time reaction 1')

df_orp_3.plot(kind='scatter', 
              x='seconds', 
              y='avg_10',
              title='avg over time reaction 3')


df_orp_2.plot(kind='scatter', 
              x='seconds', 
              y='avg_10',
              title='avg over time reaction 2')
df_orp_1.plot(kind='scatter', 
              x='seconds', 
              y='avg_10',
              title='avg over time reaction 1')


df_orp_3.plot(kind='scatter', 
              x='seconds', 
              y='sta_10',
              title='st_avg over time reaction 3')


df_orp_2.plot(kind='scatter', 
              x='seconds', 
              y='sta_10',
              title='st_avg over time reaction 2')
df_orp_1.plot(kind='scatter', 
              x='seconds', 
              y='sta_10',
              title='st_avg over time reaction 1')


```
