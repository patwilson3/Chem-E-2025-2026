```python
df = pd.read_csv("./data/orp_data/recording1.csv")
```

```python
threshold = 0.001
n_avg = 15
n_std = 15
m = len(df['orp_mV'])
```

```python
def clip_index(bottom: int=0, top : int, x:int):
    if x < bottom:
        return bottom
    elif x > top:
        return top
    else:
        return x
```

```python
means = np.zeros(m)
stds = np.zeros(m)
for i in range(m):
    means[i] =  np.mean(df['orp_mV'][clip_index(0, m, i-n_avg):clip_index(0, m, i+n_avg)])
    stds[i] = np.std(df['orp_mV'][clip_index(0, m, i-n_avg):clip_index(0, m, i+n_avg)])

df['Rolling average'] = means
df['Rollind standard deviation'] = stds
```

```python
for i in range(m):
    x = df['orp_mV']-df['Rolling average']
    y = df['Rolling standard deviation']
    if x*y < threshold:
        print(i)
```
